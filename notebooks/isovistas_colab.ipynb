```python
peca = r'''
{ COLA AQUI O JSON DA PEÇA }
'''
open('peca.json','w').write(peca)
!isovista check peca.json
```

```python
# Desenho de vistas cotado (A3, normas ISO/FEUP)
!isovista draw peca.json -o desenho
from IPython.display import Image, display
display(Image('desenho.png', width=1100))
```

```python
# Isométrico sobre reticulado (unidade = 10 mm)
!isovista iso peca.json -o iso --unit 10
display(Image('iso.png', width=900))
```

```python
# Ida-e-volta: cola o JSON escrito pela IA numa conversa NOVA, só a partir da imagem gerada
inverso = r'''
{ COLA AQUI O JSON INVERSO }
'''
open('inverso.json','w').write(inverso)
!isovista compare peca.json inverso.json
```
